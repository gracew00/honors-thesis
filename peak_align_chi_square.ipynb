```python
import pandas as pd
from scipy.stats import chisquare
```

```python
# positive control

data = {
    "col1_0": [90, 0, 0],
    "col2_1": [0, 5, 0],
    "col3_2": [0, 0, 5],
    "col4_3": [0, 0, 0],
    "col5_4": [0, 0, 0],
    "col6_5": [0, 0, 0],
    "col7_6": [0, 0, 0],
    "col8_7": [0, 0, 0],
    "col9_8": [0, 0, 0],
    "col10_9": [0, 0, 0],
}
peak_align = pd.DataFrame(data)


# create empty dictionary for conting table
contingency_table = {f'+{i}': 0 for i in range(10)}

# loop through columns in peak_align df -> get sum for each column + add to dictionary
for col in peak_align.columns:
    # getting the last digit of the column name  (e.g., "11" -> 1)
    position = int(col[-1]) 
    
    # sum the values in the column and add to the corresponding position
    contingency_table[f'+{position}'] += peak_align[col].sum()

# convert the dictionary to a dataframe
contingency_table_df = pd.DataFrame([contingency_table])

# expected data (uniform distribution)
total = contingency_table_df.sum().sum() 
num_categories = len(contingency_table_df.columns)
expected = [total / num_categories] * num_categories  

# Perform the Chi-Square test
chi2_stat, p_value = chisquare(contingency_table_df.values[0], expected)


print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-value: {p_value}")

# Interpretation
if p_value < 0.05:
    print("Significant result: The test is working as expected.")
else:
    print("Non-significant result: Check the implementation.")
```

```python
#FISCHER'S TEST
from scipy.stats import fisher_exact

# import the peak align output files
peak_align_path = "/home/ubuntu/honors_research/peak_align/+1_dinucleotide/SW_WS/merged_unique_variants_SW_WS_phase0_250bp_Output.cdt"
peak_align_df = pd.read_csv(peak_align_path, sep="\t")

# drop non-position column(s)
peak_align_df_no_name = peak_align_df.drop(columns=["NAME"])

# create empty dictionary for contingency table (+0 to +9)
contig_table = {f'+{i}': 0 for i in range(10)}

# loop through columns and sum into +position bins
for col in peak_align_df_no_name.columns:
    # get last digit of column name (e.g. "11" -> 1)
    position = int(col[-1])
    contig_table[f'+{position}'] += peak_align_df_no_name[col].sum()

# convert to Series for convenience
contig_series = pd.Series(contig_table)

# total and expected under uniform
total = contig_series.sum()
num_categories = len(contig_series)
expected_per_cat = total / num_categories

print("Observed counts per position:")
print(contig_series)
print("\nTotal:", total, "Expected per category (uniform):", expected_per_cat)

# Fisher's exact test for each position vs the rest
results = []

for pos_label, obs_pos in contig_series.items():
    obs_rest = total - obs_pos

    # expected counts under uniform
    exp_pos = expected_per_cat
    exp_rest = total - exp_pos

    # 2x2 table:
    # [ [observed at this position, observed at all other positions],
    #   [expected at this position, expected at all other positions] ]
    table = [[obs_pos, obs_rest],
             [exp_pos, exp_rest]]

    # Fisher's exact test (two-sided by default)
    odds_ratio, p_value = fisher_exact(table, alternative="two-sided")

    results.append({
        "position": pos_label,
        "obs_pos": obs_pos,
        "obs_rest": obs_rest,
        "exp_pos": exp_pos,
        "exp_rest": exp_rest,
        "odds_ratio": odds_ratio,
        "p_value": p_value,
    })

fisher_df = pd.DataFrame(results)
print("\nFisher's exact test results (position vs all others):")
print(fisher_df)
```
