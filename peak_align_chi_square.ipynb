```python
import pandas as pd
from scipy.stats import chisquare
```

```python
# positive control

data = {
    "col1_0": [90, 0, 0],
    "col2_1": [0, 5, 0],
    "col3_2": [0, 0, 5],
    "col4_3": [0, 0, 0],
    "col5_4": [0, 0, 0],
    "col6_5": [0, 0, 0],
    "col7_6": [0, 0, 0],
    "col8_7": [0, 0, 0],
    "col9_8": [0, 0, 0],
    "col10_9": [0, 0, 0],
}
peak_align = pd.DataFrame(data)


# create empty dictionary for conting table
contingency_table = {f'+{i}': 0 for i in range(10)}

# loop through columns in peak_align df -> get sum for each column + add to dictionary
for col in peak_align.columns:
    # getting the last digit of the column name  (e.g., "11" -> 1)
    position = int(col[-1]) 
    
    # sum the values in the column and add to the corresponding position
    contingency_table[f'+{position}'] += peak_align[col].sum()

# convert the dictionary to a dataframe
contingency_table_df = pd.DataFrame([contingency_table])

# expected data (uniform distribution)
total = contingency_table_df.sum().sum() 
num_categories = len(contingency_table_df.columns)
expected = [total / num_categories] * num_categories  

# Perform the Chi-Square test
chi2_stat, p_value = chisquare(contingency_table_df.values[0], expected)


print(f"Chi-Square Statistic: {chi2_stat}")
print(f"P-value: {p_value}")

# Interpretation
if p_value < 0.05:
    print("Significant result: The test is working as expected.")
else:
    print("Non-significant result: Check the implementation.")
```
