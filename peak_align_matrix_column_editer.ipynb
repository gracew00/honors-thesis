```python
import os
import glob
import pandas as pd

def drop_first_column_in_folder(folder, pattern="*.cdt", sep="\t"):
    """
    For each file matching `pattern` directly in `folder`,
    drop the first column and overwrite the file.
    No recursion into subdirectories.
    """
    folder = os.path.abspath(folder)
    print("Folder:", folder)

    files = glob.glob(os.path.join(folder, pattern))
    print("Found files:", files)

    for path in files:
        print(f"Processing {path}")
        df = pd.read_csv(path, sep=sep, header=None, dtype=str)

        if df.shape[1] < 2:
            print(f"  Skipping (only {df.shape[1]} column(s))")
            continue
            
        df = df.iloc[:, 1:]
        df.to_csv(path, sep=sep, header=False, index=False)
```
