```python
import os
import glob
import pandas as pd

def drop_yorf_column_in_folder(folder, pattern="*.cdt", sep="\t"):
    """
    For each file matching `pattern` directly in `folder`,
    drop the "YORF" column and overwrite the file.
    """
    folder = os.path.abspath(folder)
    print("Folder:", folder)

    files = glob.glob(os.path.join(folder, pattern))
    print("Found files:", files)

    for path in files:
        print(f"Processing {path}")
        df = pd.read_csv(path, sep=sep, dtype=str)

        if "YORF" not in df.columns:
            print("  No 'YORF' column, skipping")
            continue

        df = df.drop(columns=["YORF"])
        # keep header, no index
        df.to_csv(path, sep=sep, index=False)
```
